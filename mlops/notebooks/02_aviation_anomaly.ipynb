# ✈️ 02 — Aviation Flight Corridor & ADS-B Anomaly Detection
### Model: `aviation_ensemble.joblib`

This notebook focuses on aerial surveillance over the Strait of Hormuz chokepoint and Gulf airspace:
1. **ADS-B Telemetry**: Ingestion of barometric altitude, vertical rate, ground speed, and transponder squawk codes.
2. **Corridor Divergence**: Detecting deviations from commercial airway waypoints.
3. **Low-Altitude & High-Climb Rate Anomaly Scoring**.


```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import joblib

PROJECT_ROOT = Path("..").resolve()
sys.path.insert(0, str(PROJECT_ROOT / "service" / "ml-service"))

model_path = PROJECT_ROOT / "service" / "ml-service" / "models" / "aviation_ensemble.joblib"
if model_path.exists():
    aviation_bundle = joblib.load(model_path)
    print("Aviation ensemble keys:", list(aviation_bundle.keys()))
else:
    print("Aviation bundle not found.")
```

```python
# Synthetic OpenSky Flight Corridor Telemetry
rng = np.random.default_rng(42)
n_aircraft = 300

lat = rng.uniform(24.5, 27.5, n_aircraft)
lon = rng.uniform(55.0, 58.0, n_aircraft)
alt = rng.normal(32000, 4000, n_aircraft)
v_rate = rng.normal(0, 300, n_aircraft)
spd = rng.normal(450, 40, n_aircraft)

df_air = pd.DataFrame({
    "latitude": lat, "longitude": lon,
    "altitude_ft": alt, "vertical_rate_fpm": v_rate,
    "velocity_kts": spd
})
print("Air traffic test batch:")
df_air.head()
```
